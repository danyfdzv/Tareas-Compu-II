```python
#16
df_16= order_items.merge(products, on="product_id", how="left")
def costo(x):
    if x < 90:
            return 'barato'
    elif x > 120:
            return 'carito'
    else:
            return '2/3'
df_16["segmento"]= df_16["total"].apply(costo)
df_16.pivot_table(
    values="total",
    index="product_category_name",
    columns="segmento",
    aggfunc="mean"
)
```

```python
#reto jeje
df_reto = order_items.merge(
    orders,
    on="order_id",
    how="left"
)
df_reto = df_reto.merge(
    products,
    on="product_id",
    how="left"
)
df_reto["costo_flete_por_gramo"] = df_reto.apply(
    lambda r: r["freight_value"] / r["product_weight_g"] 
    if r["product_weight_g"] > 0 
    else 0,
    axis=1
)
df_reto["envio_caro"] = df_reto.apply(
    lambda r: "envio caro" if (r["freight_value"] / r["price"]) > 0.25 else "envio_bueno",
    axis=1
)
df_reto = df_reto.pivot_table(
    values=["price", "freight_value", "costo_flete_por_gramo"],
    index="product_category_name",
    columns="envio_caro",
    aggfunc={
        "price": "mean",
        "freight_value": "mean",
        "costo_flete_por_gramo": "median"
    }
)
display(df_reto)
```
