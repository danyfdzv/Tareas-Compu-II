# Manejo de columnas

Es posible seleccionar una o varias columnas:

* Una columna

* Dos columnas

* Una lista de columnas

La sintaxis general es `df[las columnas que quiero]`

### Ejemplo para una columna

Para seleccionar la columna `col15` usamos `df["col5]`

```python
df[["col5"]]
```

### Ejemplo para dos columnas

Para seleccionar dos columnas la sintaxis general es `df[["col8", "col7"]]`, la diferencia con seleccionar una columna es que en esta ocasion hay doble corchete:

```python
df[["col8", "col7"]]
```

### Ejemplo para seleccionar varias columnas

En general para seleccionar varias columnas es necesario escribir la *lista* de las columnas y pasarlas con corchetes al dataframe, por ejemplo, si queremos las columnas `col1`, `col10`, `col6` y `col8`, podemos utilizar la siguiente sintaxis

```python
columnas_seleccionadas = ["col1", "col10", "col6", "col8"]

df[columnas_seleccionadas]
```

# Filtrar Datos

Al crear una condicion sobre una columna estamos creando una mascara logica (una columna de True o False por cada fila que cumpla la condicion), por ejemplo `df["col6"]=="Sur"` crea una columna que sera True en aquellas filas donde el valor sea "Sur" y False cuando no lo sea.

```python
df["col6"] == "Sur"
```

Para condiciones numericas ocurre lo mismo, podemos utilizar operadores como >, <, ==, !=, >=, <=. Por ejemplo, para obtener las filas y columnas del dataframe que cumplan que col1 tenga un valor mayor a 500 utilizamos df[df["col1"] > 500]

```python
df[df["col1"] > 500]
```

### Filtrar por mas de una condicion

Para aplicar varios filtros al mismo tiempo, se colocan las condiciones entre parentesis y se unen mediante operadores logicos como & (and), | (or) y ~ (not). Por ejemplo:

* Las filas donde la columna 6 tenga el valor "Sur" y la columna 5 sea diferente de "B"

\* `(df["col6"] == "Sur") & (df["col5"] != "B")`

* Las filas donde la columna 5 tenga el valor "A" o la columna 6 tenga el valor "Norte"

\* `(df["col5"] == "A") | (df["col6"] == "Norte")`

```python
df[(df["col6"] == "Sur") & (df["col5"] != "B")]
```

# Ordenar datos

Para ordenar los datos de menor a mayor podemos usar df.sort_values(). Por ejemplo, si queremos ordenar la columna 1, usamos df.sort_values("col1").

```python
df.sort_values("col1")
```

```python
df.sort_values(["col1", "col5"])
```

# Estadisticas descriptivas

Para obtener diferentes datos de una columna, primero seleccionamos la columna y usamos el metodo que necesitemos. Por ejemplo:

.mean() calcula la media o promedio de la columna 1.

.median() calcula la mediana de la columna 1.

.min() muestra el valor mas pequeño de la columna 1.

.max() muestra el valor mas grande de la columna 1.

.std() muestra que tan dispersos estan los datos con respecto a la media.

```python
print("promedio:")
display(df["col1"].mean())
```

```python
print("mediana:")
display(df["col1"].median())
```

```python
print("mínimo:")
display(df["col1"].min())

print("máximo:")
display(df["col1"].max())
```

## Trabajar con variables categoricas

* value_counts() sirve para saber cuantas veces se repite cada elemento.

* unique() muestra los valores diferentes que hay en una columna.

* nunique() cuenta cuantos valores unicos hay en una columna.

```python
df["col1"].value_counts()
```

```python
df["col1"].unique()
```

```python
df["col1"].nunique()
```

## Agrupar datos 
df.groupby()

Con este metodo podemos agrupar datos de una o varias columnas y realizar diferentes acciones, como sacar promedios o contar elementos por categoria.

Para usar groupby, escribimos:

df.groupby("col10")[["col1", "col5"]]

```python
df.groupby("col10")["col1"].mean()
```

```python
df.groupby(["col10", "col5"])[["col1", "col3"]].mean()
```

```python
df.groupby(["col10", "col5"])[["col1", "col3"]].count()
```
